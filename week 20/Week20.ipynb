```python
iris = sns.load_dataset('iris')
```

```python
iris.head()
```

```python
sns.pairplot(iris, markers = 'o', hue = 'species')
# sns.pairplot?
plt.show()
```

```python
# plt.figure(figsize = (10,6))
iris_long = pd.melt(
    iris,
    id_vars="species",                # keep species as identifier
    var_name="feature",                # column name for features
    value_name="value"                  # column name for values
)
sns.lineplot(data = iris_long,x = 'species', y = 'value',style = 'feature', marker = "o")
```

```python
sns.boxplot(data =iris_long, x = 'feature', y = 'value', hue = 'species')
```

```python
plt.figure(figsize = (10,6))
sns.violinplot(x = 'feature',y = 'value',data =iris_long, hue = 'species')
```

```python
sns.relplot(y = 'value', x = 'feature', data =iris_long, hue = 'species', style = 'feature')   
```
