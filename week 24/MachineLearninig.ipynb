```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression as LR
```

```python
x = np.array([2,3,4,5,6,7]).reshape(-1,1)  #hours
y = np.array([50,60,65, 80,85,90])  #marks
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size= 0.2, random_state= 42)
```

```python
x_train, x_test, y_train, y_test
```

```python
model = LR()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
sns.scatterplot(x = x.flatten(), y = y)
plt.plot(x, model.predict(x),color = 'r')
new_data = np.array([2.5,3.5,9]).reshape(-1,1)
new_data_pred = model.predict(new_data)
sns.scatterplot(x = new_data.flatten() , y = new_data_pred)
```

```python
diabeties = load_diabetes()
diabeties
```

```python
dataset = pd.DataFrame(diabeties.data, columns = diabeties.feature_names)
```

```python
dataset.head()
dataset
```

```python
y = diabeties.target
y
```

```python
x_d_train, x_d_test, y_d_train, y_d_test = train_test_split(dataset,y, test_size=0.2, random_state=42)
```

```python
x_d_train.shape
```

```python
x_d_test.shape
```

```python
model_d = LR()
```

```python
model_d.fit(x_d_train,y_d_train)
```

```python
d_train_pred = model_d.predict(x_d_train)
plt.scatter(x= x_d_train['s6'], y=y_d_train, alpha=0.3)
plt.scatter(x= x_d_train['s6'], y=d_train_pred, alpha= 0.3)
plt.show()

for col in dataset.columns:
    sns.regplot(x = dataset[col], y = y, scatter_kws={'alpha':0.5})
    plt.show()
```

```python
d_test_pred = model_d.predict(x_d_test)
plt.scatter(x= x_d_test['s6'], y=y_d_test, alpha=0.3)
plt.scatter(x= x_d_test['s6'], y=d_test_pred, alpha= 0.3)
```
